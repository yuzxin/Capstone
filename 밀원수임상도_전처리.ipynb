```python
import pandas as pd
import glob

# 1. 파일 불러오기
files = glob.glob('/content/drive/MyDrive/캡스톤/임상도소스데이터/*_loc.csv')  # 업로드 위치에 맞게 수정

df_list = []
for f in files:
    df = pd.read_csv(f, encoding='cp949', low_memory=False)
    df_list.append(df)

df = pd.concat(df_list, ignore_index=True)

print("전체 데이터:", df.shape)
```

```python
df.info()
```

```python
df.head()
```

```python
# 필요한 컬럼만 선택

df = df[[
    'EMD_CD',
    'EMD_NM',
    'KOFTR_NM',
    'Shape_Area'
]]

df.columns = [
    'emd_cd',
    'emd_nm',
    'species',
    'area_m2'
]
```

```python
# 결측치 및 깨진 데이터 제거

df = df.dropna(subset=['emd_cd', 'emd_nm'])

# 숫자로 깨진 행 제거 (예: 5.18E+18)
df = df[df['emd_nm'].str.contains('[가-힣]', na=False)]

print("정제 후:", df.shape)
```

```python
# 헥타르 단위의 면적 컬럼 생성
df['area_ha'] = df['area_m2'] / 10000
df.head()
```

```python
# 밀원수 리스트 불러오기

melliferous_df = pd.read_excel('/content/drive/MyDrive/캡스톤/밀원자원수목류_수정.xlsx')

melliferous_list = melliferous_df['밀원수명'].unique().tolist()

print("밀원수 개수:", len(melliferous_list))
```

```python
df = df[df['species'].isin(melliferous_list)]
df.head()
```

```python
acacia_df = df[df['species'] == '아까시나무']
display(acacia_df.head())
```

```python
# 법정동 코드로 시도, 시군구 컬럼 추가
law = pd.read_csv('/content/drive/MyDrive/캡스톤/법정동코드 전체자료.txt', sep='\t', encoding='cp949')
law.columns = ['code', 'name', 'status']
law = law[law['status'] == '존재']

law['code'] = law['code'].astype(str).str.zfill(8)

law['sido_cd'] = law['code'].str[:2]
law['sigungu_cd'] = law['code'].str[:5]

law['sido'] = law['name'].str.split().str[0]
law['sigungu'] = law['name'].str.split().str[1]

sido = law[['sido_cd', 'sido']].drop_duplicates()
sigungu = law[['sigungu_cd', 'sigungu']].drop_duplicates()

df['emd_cd'] = df['emd_cd'].astype(str).str.zfill(8)
df['sido_cd'] = df['emd_cd'].str[:2]
df['sigungu_cd'] = df['emd_cd'].str[:5]

df = df.merge(sido, on='sido_cd', how='left')
df = df.merge(sigungu, on='sigungu_cd', how='left')
```

```python
df.head()
```

```python
# 컬럼명 변경
df = df.rename(columns={'emd_nm':'emd'})

df.head()
```

```python
# 컬럼 순서 재배열
df = df[[
    'sido',
    'sigungu',
    'emd',
    'sido_cd',
    'sigungu_cd',
    'emd_cd',
    'species',
    'area_m2',
    'area_ha'
]]

df.head()
```

```python
df = df.sort_values(['sido', 'sigungu', 'emd'])
df.head()
```
