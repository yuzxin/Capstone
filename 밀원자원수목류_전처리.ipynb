# 밀원자원수목류_전처리

```python
import pandas as pd
import numpy as np
import re
```

```python
# 데이터 불러오기
df = pd.read_excel("/content/drive/MyDrive/캡스톤/밀원자원수목류.xlsx")

print(df.shape)
display(df.head())
```

```python
# 컬럼명 정리
df.columns = (
    df.columns
      .str.strip()
      .str.replace("\n","",regex=False)
)
df.head()
```

```python
# 분석에 사용하지 않을 컬럼 제거
drop_cols = [
    "자원종류",
    "개화량/150m²",
    "잠재적 꿀 생산량(g/3m²)"
]

df = df.drop(columns=drop_cols)

print(df.columns)
```

```python
# 문자열 공백 제거
text_cols = [
    "밀원수명",
    "개화시기기준지역",
    "꿀 생산량 등급"
]

for c in text_cols:
    df[c] = (
        df[c]
        .astype(str)
        .str.strip()
        .replace("nan", np.nan)
    )
df.head()
```

```python
# 숫자형 변환
num_cols = [

    "개화량/그루",

    "화밀분비량(μ/flower)",

    "유리당 함량(mg/flower)",

    "잠재적 꿀 생산량(g/그루)",

    "잠재적 꿀 생산량(kg/ha)",

    "꿀 생산량 등급"
]

for c in num_cols:
    df[c] = pd.to_numeric(df[c], errors="coerce")

df.head()
```

```python
# 개화시기 분리
df["개화시기"] = (
    df["개화시기"]
      .astype(str)
      .str.replace(" ","")
)

split = df["개화시기"].str.split("-", expand=True)

df["개화_시작"] = split[0]
df["개화_종료"] = split[1]

# 날짜를 datetime으로 변환
YEAR = 2025

df["개화_시작_날짜"] = pd.to_datetime(
    YEAR.astype(str) if False else str(YEAR) + "." + df["개화_시작"],
    format="%Y.%m.%d",
    errors="coerce"
)

df["개화_종료_날짜"] = pd.to_datetime(
    str(YEAR) + "." + df["개화_종료"],
    format="%Y.%m.%d",
    errors="coerce"
)

# 연중 일수(DOY) 생성
df["개화_시작_DOY"] = df["개화_시작_날짜"].dt.dayofyear

df["개화_종료_DOY"] = df["개화_종료_날짜"].dt.dayofyear

# 개화기간 계산
df["개화기간"] = (
    df["개화_종료_DOY"] -
    df["개화_시작_DOY"] + 1
)
```

```python
# 개화 중심일
df["개화중심_DOY"] = (
    (df["개화_시작_DOY"] + df["개화_종료_DOY"]) / 2
).round().astype("Int64")

# 개화기간
df["개화기간"] = (
    df["개화_종료_DOY"] -
    df["개화_시작_DOY"] + 1
)

# 개화 시작 월
df["개화월"] = df["개화_시작_날짜"].dt.month

drop_cols = [
    "개화_시작_날짜",
    "개화_종료_날짜"
]

df = df.drop(columns=drop_cols)

print(df.columns)
```

```python
# 대표 개화일 기준 정렬
df = df.sort_values("개화중심_DOY").reset_index(drop=True)

# 같은 대표개화일은 같은 순위 부여
df["개화순서"] = (
    df["개화중심_DOY"]
      .rank(method="dense")
      .astype(int)
)
```

```python
# 결측치 확인
missing = (
    df.isnull()
      .sum()
      .to_frame("결측수")
)

missing["결측률(%)"] = (
    missing["결측수"] /
    len(df) * 100
).round(2)

display(missing)
```

```python
# 컬럼 재정렬
column_order = [
    "밀원수명",
    "기타",
    "개화시기기준지역",

    "개화시기",
    "개화_시작",
    "개화_종료",

    "개화_시작_DOY",
    "개화_종료_DOY",
    "개화중심_DOY",
    "개화순서",
    "개화기간",
    "개화월",

    "개화량/그루",
    "화밀분비량(μ/flower)",
    "유리당 함량(mg/flower)",

    "잠재적 꿀 생산량(g/그루)",
    "잠재적 꿀 생산량(kg/ha)",
    "꿀 생산량 등급"
]

df = df[column_order]
```

```python
# 컬럼명 영어로 변경
df = df.rename(columns={
    "밀원수명": "species_name",
    "기타": "add_info",
    "개화시기기준지역": "bloom_region",

    "개화시기": "bloom_period",
    "개화_시작": "bloom_start",
    "개화_종료": "bloom_end",

    "개화_시작_DOY": "bloom_start_doy",
    "개화_종료_DOY": "bloom_end_doy",
    "개화중심_DOY": "bloom_mid_doy",
    "개화순서": "bloom_order",
    "개화기간": "bloom_duration",
    "개화월": "bloom_month",

    "개화량/그루": "flower_amount_per_tree",
    "화밀분비량(μ/flower)": "nectar_per_flower",
    "유리당 함량(mg/flower)": "sugar_per_flower",

    "잠재적 꿀 생산량(g/그루)": "honey_yield_per_tree",
    "잠재적 꿀 생산량(kg/ha)": "honey_yield_per_ha",
    "꿀 생산량 등급": "honey_grade"
})
df.head()
```
